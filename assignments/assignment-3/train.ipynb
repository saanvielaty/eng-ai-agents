```python
from roboflow import Roboflow

rf = Roboflow(api_key="a2pqX0EdrrjwfeX6yrN7")  # paste your roboflow key
project = rf.workspace("drone-detection-g4d3g").project("drone-detection-a1tsf")
version = project.version(6)
dataset = version.download("yolov8")
```

```python
import yaml

yaml_path = "/content/Drone-Detection-6/data.yaml"

with open(yaml_path, "w") as f:
    yaml.dump({
        "names": ["drones"],
        "nc": 1,
        "train": "/content/Drone-Detection-6/train/images",
        "val": "/content/Drone-Detection-6/valid/images",
        "test": "/content/Drone-Detection-6/test/images"
    }, f)

print("data.yaml fixed!")
```

```python
from ultralytics import YOLO

model = YOLO("yolov8s.pt")

results = model.train(
    data="/content/Drone-Detection-6/data.yaml",
    epochs=20,
    imgsz=640,
    batch=16,
    name="drone_detector",
    project="/content/runs",
    patience=5,
    device="cuda",      # GPU on Colab
    workers=2,
    verbose=True
)

print("Training complete!")
print("Best model at:", results.save_dir)
```
